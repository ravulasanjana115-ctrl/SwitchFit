```python
import numpy as np
import pandas as pd
import plotly.express as px
from textblob import TextBlob

df = pd.read_csv("netflix_titles.csv.zip")
```

```python
x = df.groupby(['rating']).size().reset_index(name='counts')

print(x)
```

```python
pieChart = px.pie(
    x,
    values='counts',
    names='rating',
    title='Distribution of content ratings on Netflix'
)

pieChart.show()
```
